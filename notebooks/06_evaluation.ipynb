## 1. Evaluation Methodology
To strictly evaluate cross-subject generalization on the DEAP dataset, all models are evaluated on the **independent test participants** ($s28$ to $s32$):
- **Mean Absolute Error (MAE):** Average absolute deviation from ground-truth continuous ratings:
  $$\text{MAE} = \frac{1}{N} \sum_{i=1}^N |y_i - \hat{y}_i|$$
- **Root Mean Squared Error (RMSE):** Penalizes larger affective prediction deviations:
  $$\text{RMSE} = \sqrt{\frac{1}{N} \sum_{i=1}^N (y_i - \hat{y}_i)^2}$$
- **Coefficient of Determination ($R^2$):** Proportion of variance in emotional ratings explained by EEG features:
  $$R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$$
- **Pearson Correlation Coefficient ($r$):** Linear alignment tracking between true and predicted affective trajectories:
  $$r = \frac{\sum (y_i - \bar{y})(\hat{y}_i - \bar{\hat{y}})}{\sqrt{\sum (y_i - \bar{y})^2 \sum (\hat{y}_i - \bar{\hat{y}})^2}}$$

```python
# Load Model Metrics Tables
metrics_path = Path(RESULTS_DIR) / "metrics"
metric_files = list(metrics_path.glob("*_test_metrics.csv"))

summary_tables = {}
for f in metric_files:
    model_name = f.stem.replace("_test_metrics", "").upper()
    summary_tables[model_name] = pd.read_csv(f)
    print(f"\n=== {model_name} Test Performance ===")
    print(summary_tables[model_name].to_string(index=False))
    
if not metric_files:
    print("No saved metric files found yet. Run training pipeline (notebooks 03-05 or colab_train.py).")
```
